## Question 1: Reverse a String

**Problem:** Write a program that reverses a given string without changing the original variable.

**Logic:** Python slicing with a step of `-1` visits the characters from the end to the beginning.

**Sample output:** `nohtyP`

```python
text = "Python"
reversed_text = text[::-1]
print(reversed_text)
```

## Question 2: Check Whether a String Is a Palindrome

**Problem:** Determine whether a phrase reads the same forward and backward, ignoring spaces, punctuation, and letter case.

**Logic:** Keep only alphanumeric characters, convert them to lowercase, and compare the normalized text with its reverse.

**Sample output:** `True`

```python
phrase = "A man, a plan, a canal: Panama"
normalized = "".join(character.lower() for character in phrase if character.isalnum())
is_palindrome = normalized == normalized[::-1]
print(is_palindrome)
```

## Question 3: Count Character Frequencies

**Problem:** Count how many times each character appears in a string. For this example, count letters and ignore spaces.

**Logic:** Visit each character and use a dictionary to store and increment its count.

**Sample output:** `{'b': 1, 'a': 3, 'n': 2}`

```python
text = "banana"
character_counts = {}
for character in text:
    character_counts[character] = character_counts.get(character, 0) + 1
print(character_counts)
```

## Question 4: Find the Largest Number in a List

**Problem:** Find the largest value in a non-empty list of numbers.

**Logic:** Start with the first value as the current maximum, then replace it whenever a larger value is found.

**Sample output:** `42`

```python
numbers = [17, 3, 42, 8, 29]
largest = numbers[0]
for number in numbers[1:]:
    if number > largest:
        largest = number
print(largest)
```

## Question 5: Remove Duplicates While Preserving Order

**Problem:** Create a new list containing each value from the input list only once, keeping its first-occurrence order.

**Logic:** Add a value only if it has not appeared in the result yet.

**Sample output:** `[1, 2, 3, 4]`

```python
values = [1, 2, 2, 3, 1, 4]
unique_values = []
for value in values:
    if value not in unique_values:
        unique_values.append(value)
print(unique_values)
```

## Question 6: FizzBuzz

**Problem:** For numbers 1 through 15, print `Fizz` for multiples of 3, `Buzz` for multiples of 5, `FizzBuzz` for multiples of both, and the number otherwise.

**Logic:** Check divisibility by both numbers first, then check each individual condition.

**Sample output:** `1, 2, Fizz, 4, Buzz, Fizz, 7, 8, Fizz, Buzz, 11, Fizz, 13, 14, FizzBuzz` (one item per line)

```python
for number in range(1, 16):
    if number % 3 == 0 and number % 5 == 0:
        result = "FizzBuzz"
    elif number % 3 == 0:
        result = "Fizz"
    elif number % 5 == 0:
        result = "Buzz"
    else:
        result = number
    print(result)
```

## Question 7: Calculate a Factorial with a Function

**Problem:** Write a function that returns the factorial of a non-negative integer. By definition, `0!` is `1`.

**Logic:** Multiply all integers from 2 through the input value, starting from 1.

**Sample output:** `120`

```python
def factorial(number):
    if number < 0:
        raise ValueError("Factorial is only defined here for non-negative integers.")
    result = 1
    for value in range(2, number + 1):
        result *= value
    return result

print(factorial(5))
```

## Question 8: Generate the Fibonacci Sequence

**Problem:** Write a function that returns the first `n` terms of the Fibonacci sequence, beginning with 0 and 1.

**Logic:** Repeatedly append the current term and update the pair to the next two terms.

**Sample output:** `[0, 1, 1, 2, 3, 5, 8]`

```python
def fibonacci(count):
    sequence = []
    first, second = 0, 1
    for _ in range(count):
        sequence.append(first)
        first, second = second, first + second
    return sequence

print(fibonacci(7))
```

## Question 9: Find Common Values in a List and a Tuple

**Problem:** Return the values that occur in both a list and a tuple, without repeating values and in the order they first occur in the list.

**Logic:** Convert the tuple to a set for membership checks, then traverse the list and append matching values once.

**Sample output:** `[2, 3]`

```python
list_values = [1, 2, 3, 2, 4]
tuple_values = (3, 5, 2, 6)
tuple_lookup = set(tuple_values)
common_values = []
for value in list_values:
    if value in tuple_lookup and value not in common_values:
        common_values.append(value)
print(common_values)
```

## Question 10: Create a Simple Bank Account Class

**Problem:** Define a `BankAccount` class with a starting balance, a deposit method, and a withdrawal method. Reject negative deposits and withdrawals larger than the current balance.

**Logic:** Store the balance as an instance attribute and update it only when each transaction is valid.

**Sample output:** `Balance: 125`

```python
class BankAccount:
    def __init__(self, opening_balance=0):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative.")
        self.balance = opening_balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive.")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive.")
        if amount > self.balance:
            raise ValueError("Insufficient funds.")
        self.balance -= amount

account = BankAccount(100)
account.deposit(50)
account.withdraw(25)
print(f"Balance: {account.balance}")
```
